```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

```python
outputs = pd.read_csv(r"Z:\Marina\3D_Permeability_Vasculature_From_Bel\outputs\combined_outputs.csv")
```

```python
outputs["sa/v"] = outputs["vasculature_surface_area_um2"]/outputs["vascular_volume_um3"]
```

```python
plot_df = outputs[(outputs["p_cm/s"] < 0.9e-5) & (outputs["p_cm/s"]>0)].copy()

fig, ax = plt.subplots(ncols=2, figsize=(8, 5), sharey=True)
sns.scatterplot(data=plot_df, x="final_gel_intensity", y="p_cm/s", hue="condition", ax=ax[0], s=7, legend=False)
sns.scatterplot(data=plot_df, x="sa/v", y="p_cm/s", hue="condition", ax=ax[1], s=7, legend=True)

handles, labels = ax[1].get_legend_handles_labels()
if ax[1].legend_ is not None:
    ax[1].legend_.remove()
fig.legend(handles, labels, title="condition", loc="right",  ncol=1, bbox_to_anchor = (1.15,0.5))

plt.tight_layout()
plt.savefig("issues_with_vasculature_intensity.pdf")
plt.savefig("issues_with_vasculature_intensity.png")
```

```python
plot_df["log_p_cm/s"] = np.log10(plot_df["p_cm/s"] )

fig, ax = plt.subplots(ncols=2, figsize=(8, 5), sharey=True)
sns.scatterplot(data=plot_df, x="final_gel_intensity", y="log_p_cm/s", hue="condition", ax=ax[0], s=7, legend=False)
sns.scatterplot(data=plot_df, x="sa/v", y="log_p_cm/s", hue="condition", ax=ax[1], s=7, legend=True)

handles, labels = ax[1].get_legend_handles_labels()
if ax[1].legend_ is not None:
    ax[1].legend_.remove()
fig.legend(handles, labels, title="condition", loc="right",  ncol=1, bbox_to_anchor = (1.15,0.5))

plt.tight_layout()
plt.savefig("issues_with_vasculature_intensity_log.pdf")
plt.savefig("issues_with_vasculature_intensity_log.png")



```
